**Editing: Inspecting and Managing Documents in a Vectorstore**

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document
```

```python
# Embedding Model Class (Assigning Similarity Scores)
embedding = OpenAIEmbeddings(
    model="text-embedding-3-small",
    chunk_size=1,          # <= key: send 1 text at a time
    request_timeout=180,   # longer timeout
    max_retries=12,        # retry on drops
)
```

```python
# Editting and Pulling Vector Files
vectorstore_from_directory = Chroma(embedding_function = embedding, persist_directory = "./intro-to-ds-lectures")
```

```python
vectorstore_from_directory.get('823e5846-738f-4b33-91ae-4719478fe1e5',include=["embeddings"])
```

*Adding Document, Updating, and Deleting for the Vector Database*

```python
added_document = Document(page_content="Analysis vs Analytics Alright! So… Let’s discuss the not-so-obvious differences between the terms analysis and analytics. Due to the similarity of the words, some people believe they share the same meaning, and thus use them interchangeably. Technically, this isn’t correct. There is, in fact, a distinct difference between the two. And the reason for one often being used instead of the other is the lack of a transparent understanding of both,",                        
                          metadata = {'Course Title': 'Introduction to Data and Data Science',
                                       'Lecture Title': 'Analysis vs Analytics'},
)
```

```python
vectorstore_from_directory.add_documents([added_document])
```

```python
vectorstore_from_directory.get('85fa475f-7f19-443e-9c3c-cf9259998510',include=["embeddings","documents","metadatas"])
```

```python
updated_document = Document(page_content = 'Great! We hope we gave you a good idea about the level of applicability of the most frequently used programming and software tools in the field of data science. Thank you for watching!',
                            metadata={'Course Title':'Introduction to Data and Data Science',
                                      'Lecture Title':'Programming Languages & Software Employed in Data Science - All the Tools'}
)
```

```python
vectorstore_from_directory.update_document(document_id="85fa475f-7f19-443e-9c3c-cf9259998510",
                                           document = updated_document)
```

```python
vectorstore_from_directory.get('85fa475f-7f19-443e-9c3c-cf9259998510',include=["embeddings","documents","metadatas"])
```

```python
vectorstore_from_directory.delete('85fa475f-7f19-443e-9c3c-cf9259998510')
```
