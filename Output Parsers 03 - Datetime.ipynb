```python
%load_ext dotenv
%dotenv

from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain.output_parsers.datetime import DatetimeOutputParser

date_output_parser = DatetimeOutputParser()

chat = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0,
    max_tokens=100,
    seed=365
)

message_h = HumanMessage(
    content=f"""When was the Danish poet Piet Hein born?
{date_output_parser.get_format_instructions()}
Return only the date/time in that exact format."""
)

response = chat.invoke([message_h])
print("RAW:", response.content)

response_parsed = date_output_parser.parse(response.content)
print("PARSED:", response_parsed, type(response_parsed))
```
