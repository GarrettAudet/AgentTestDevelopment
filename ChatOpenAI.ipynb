**ChatOpenAI (LangChain Wrapper for OpenAI)**

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
```

```python
chat = ChatOpenAI(model_name = 'gpt-4', 
                  model_kwargs = {'seed':365}, 
                  temperature = 0,
                 max_tokens = 100)

# ChatOpenAI (LangChain Wrapper, Allows me to Clal OpenAI Models with LangChain abstractions like chains, prompts, tools, and agents).
#kwargs, dict of parameters
#temperature, reduces randomness of answers 
```

```python
# System and Human Messages for OpenAI in Langchain
message_s = SystemMessage(content = ''' You are Marv, a chatbot that reluctantly answers questions with sarcastic respones ''') 
message_h = HumanMessage(content = ''' I've recently adopted a cat. Can you suggest some cat names? ''') 

# System Message: Stores Chatbot Instructions
# Human Message: Storing User Requests
# AI Message: Assistant-role Message in OpenAI's API
# Response Variable: AI Messages (Generated by Model)
```

```python
response = chat.invoke([message_s,message_h])
```

```python
print(response.content)
```

**ChatOpenAI (LangChain Wrapper - Few Shot Prompting Example)**

```python
chat = ChatOpenAI(model_name = 'gpt-4', 
                  model_kwargs = {'seed':365}, 
                  temperature = 0,
                 max_tokens = 100)
```

```python
# Few Shot Prompting Example
# Example 1
message_h_dog = HumanMessage(content = ''' I've recently adopted a dog. Can you suggest some dog names? ''') 
message_ai_dog = AIMessage(content = ''' Oh, absolutely. Because I'm sure your dog is just waiting to be named by a chatbot. 
How about "Bark Twain" if it's a literary hound, or "Sir Wag-a-lot" if it's a knight in shining armor? Maybe "Bark Zuckerberg" if it's into social media, or "Bark Obama" if it's a political pup. 
And if it's a bit of a diva, "Barkonce" could work. You''') 

# Example 2
message_h_cat = HumanMessage(content = ''' I've recently adopted a cat. Can you suggest some cat names? ''') 
message_ai_cat = AIMessage(content = ''' Oh, absolutely. Because I'm sure your cat is just waiting to ignore whatever name you 
decide to give it. How about "Sir Scratches-a-lot", "Madam Meow", "Captain Furball", or "Miss Hiss-a-lot"? Or if you 
want to be really original, you could go with "Cat". I'm sure no one's thought of that one before.''') 

message_h_fish = HumanMessage(content = ''' I've recently adopted a cat. Can you suggest some cat names? ''') 
```

```python
response = chat.invoke([message_h, message_ai_dog, message_h_cat, message_ai_cat,message_h_fish])
```

```python
print(response.content)
```
