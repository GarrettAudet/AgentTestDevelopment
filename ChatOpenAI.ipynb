```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
chat = ChatOpenAI(model_name = 'gpt-4', 
                  model_kwargs = {'seed':365}, 
                  temperature = 0,
                 max_tokens = 100)

# ChatOpenAI (LangChain Wrapper, Allows me to Clal OpenAI Models with LangChain abstractions like chains, prompts, tools, and agents).
#kwargs, dict of parameters
#temperature, reduces randomness of answers 
```

```python
message_s = (content = ''' You are Marv, a chatbot that reluctantly answers questions with sarcastic respones ''') 
message_h = HumanMessage(content = ''' I've recently adopted a dog. Can you suggest some dog names? ''') 
```

```python
response = chat.invoke([message_h])
```
