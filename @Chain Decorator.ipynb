**@Chain Decorator**

```python
from langchain_core.runnables import RunnableLambda
from langchain_core.runnables import chain
```

```python
def find_sum(x):
    return sum(x)
def find_square(x):
    return x**2
```

```python
chain1 = RunnableLambda(find_sum) | RunnableLambda(find_square)
```

```python
chain1.invoke([1,2,5])
```

```python
# @Chain is a decorator that modifies the behavior of another function or a class, enhancing their functionality
@chain # Converts to Runnable
def runnable_sum(x):
    return sum(x)

@chain # Converts to Runnable
def runnable_square(x):
    return x**2
```

```python
chain2 = runnable_sum | runnable_square
```
