**Setting the API Key as an Environment Variable**

```python
%load_ext dotenv
%dotenv
```

```python
import os 
import openai
```

```python
openai.api_key = os.getenv("OPENAI_API_KEY")
if not openai.api_key:
    raise ValueError("OPENAI_API_KEY not found")
print("API key loaded:", bool(openai.api_key))
```

***Example 1: Sarcastic Chatbot***

```python
client = openai.OpenAI()
```

```python
completion = client.chat.completions.create(model = 'gpt-4',
                                            messages = [{'role':'system',
                                                        'content':'''You are Marv, a chatbot that reluctantly
                                                        answers questions with sarcastic responses. '''},
                                                        {'role':'user',
                                                         'content':'''Explain what a Black Hole is '''}], 
                                            max_tokens = 250, temperature = 0, seed = 365, stream = True)

# Max Tokens, Limits Tokens per Response
# Temperature, Reduces Randomness of Answers (Better for Academic Answers
# Seed, Model tries to make the same sampling choices each run, so you get (usually) very similar outputs (consistency) 
```

```python
completion
```

```python
print(completion.choices[0].message.content)
```

***Example 2: Tweet Classifier***

```python
completion = client.chat.completions.create(model = 'gpt-4',
                                            messages = [{'role':'system',
                                                        'content':''' You will be provided with a tweet, and your task is to classify its
                                                        sentiment as positive, neutral, or negative'''},
                                                        {'role':'user',
                                                         'content':'''I loved that new movie at the Cinema! '''},
                                                        {'role':'assistant', 
                                                         'content':'''Neutral.'''}])
```
