**Creating a Prompt, Leveraging a Model, and Using an Output Parser**

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
list_instructions = CommaSeparatedListOutputParser().get_format_instructions()
```

```python
list_instructions
```

```python
chat_template = ChatPromptTemplate.from_messages(['human',"I've recently adopted a {pet}. Could you suggest three {pet} names? \n" 
                                                  + list_instructions])
```

```python
print(chat_template.messages[1].prompt.template)
```

```python
chat = ChatOpenAI(model_name = 'gpt-4',
                  model_kwargs = {'seed':365},
                  temperature = 0,
                  max_tokens = 100)
```

```python
list_output_parser = CommaSeparatedListOutputParser()
```

```python
chat_template_result = chat_template.invoke({'pet':'dog'})
```

```python
chat_result = chat.invoke(chat_template_result)
```

```python
list_output_parser.invoke(chat_result)
```

```python
# Pipe Symbol Links Elements in the Expression Language, Showing the Output will be the Input for the Next Component
# Takes the Output from the Left and Feeds to the Right
# Chat_template (instance of the ChatPromptTemplate class, chat is my model isntance, list_output_parser is my parser instance that tranforms outputs
# Langchain flow is: template instance builds prompt/messages, model instance generates text/message, parser instance converts that output to list[str]

chain = chat_template | chat | list_output_parser
```

```python
chain.invoke({'pet':'dog'})
```
