**Retrieval: Similarity Search Method**

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document
```

```python
# Embedding Model Class (Assigning Similarity Scores)
embedding = OpenAIEmbeddings(
    model="text-embedding-3-small",
    chunk_size=1,          # <= key: send 1 text at a time
    request_timeout=180,   # longer timeout
    max_retries=12,        # retry on drops
)
```

```python
# Vector Store 
vectorstore = Chroma(embedding_function = embedding, persist_directory = "./intro-to-ds-lectures")
```

```python
len(vectorstore.get()['documents'])
```

```python
# Can Pick Retriever (search_type_), and keywords
retriever = vectorstore.as_retriever(search_type='mmr',
                                     search_kwags={
                                         'k':3,
                                         'lambda_mult':0.7
                                     })
```

```python
retriever
```

```python
question = "what software does data scientists use?"
```

```python
retrieved_docs = retriever.invoke(question)
```

```python
for i in retrieved_docs:
    print(f"{i.page_content}")
```
