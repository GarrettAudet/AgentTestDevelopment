**Piping Chains and the RunnablePassthrough Class**

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
```

```python
chat_template_tools = ChatPromptTemplate.from_template('''What are the five most important tools a {job title} needs?
                                                       Answer only by listing the tools. ''') 

chat_template_strategy = ChatPromptTemplate.from_template('''Consider the tools provided, develop a strategy for effectively learning
                                                        and mastering them: {tools}''')
```

```python
chat_template_tools
```

```python
chat = ChatOpenAI(model_name = 'gpt-4',
                  model_kwargs = {'seed':365},
                  temperature = 0,
                  max_tokens = 100)
```

```python
string_parser = StrOutputParser()
```

```python
chain_tools = chat_template_tools | chat | string_parser
chain_strategy = chat_template_strategy | chat | string_parser
```

```python
print(chain_tools.invoke({'job title':'data scientist'}))
```

```python
print(chain_strategy.invoke({'tools':'''
1. Python
2. R Programming
3. SQL
4. Tableau
5. Hadoop
'''}))
```
