*Import Relevant Classes and Functions*

```python
from langgraph.graph import START, END, StateGraph
from typing_extensions import TypedDict # Structured dictionary with fixed keys 
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import HumanMessage, BaseMessage
from langchain_core.runnables import Runnable
from collections.abc import Sequence # Sequence is an ordered collection of items  (list, tuple, str)
```

*Define the State*

```python
class State(TypedDict):
    messages: Sequence[BaseMessage] # Messages should be an ordered collection of BaseMessage objects 

# This is saying that the Class is a TypedDict inheritance where there is a value of 'messages' that should have 1 or more BaseMessage objects 
```

```python
state = State(messages = [HumanMessage("Could you tell me a grook by Piet Hein?")])
```

```python
state["messages"][0].pretty_print()
```

*Define the Nodes*

```python
chat = ChatOpenAI(model='gpt-4o',
                  seed = 365,
                  temperature = 0,
                  max_completion_tokens = 100)
```

```python
response = chat.invoke(state['messages'])
```

```python
response.pretty_print()
```

```python
def chatbot(state: State) -> State:
    print(f"\n------> ENTERING chatbot:")
    response = chat.invoke(state['messages'])
    response.pretty_print()
    return State(messages = [response])

# Ingests State, Reads Current Statem Calls the Model with the Messages, Returns an Updated State to Feed into the Next Node 
```

```python
chatbot(state)
```
