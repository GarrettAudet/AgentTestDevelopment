**Indexing: Document Splitting with Character Text Splitting**

```python
from langchain_community.document_loaders import Docx2txtLoader
from langchain_text_splitters.character import CharacterTextSplitter
```

```python
# Loads Files
loader_pdf = Docx2txtLoader("RAGFiles/Introduction_to_Data_and_Data_Science.docx")
pages = loader_pdf.load()
for i in range(len(pages)):
    pages[i].page_content = ' '.join(pages[i].page_content.split())
```

```python
len(pages[0].page_content)
```

```python
# Separator separates words
char_splitter = CharacterTextSplitter(separator = "", 
                                      chunk_size = 500, 
                                      chunk_overlap = 50)
```

```python
pages_char_split = char_splitter.split_documents(pages)
```

```python
# 17 Chunks 
len(pages_char_split)
```
