```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
chat_template = ChatPromptTemplate.from_messages(['human', """I've recently adopted a {pet} which is a {breed}. 
                                                Could you suggest several training tips? \n"""])
```

```python
chat = ChatOpenAI(model_name = 'gpt-4',
                  model_kwargs = {'seed':365},
                  temperature = 0,
                  max_tokens = 100)
```

```python
# Pipe Symbol Links Elements in the Expression Language, Showing the Output will be the Input for the Next Component
# Takes the Output from the Left and Feeds to the Right
# Chat_template (instance of the ChatPromptTemplate class, chat is my model isntance, list_output_parser is my parser instance that tranforms outputs
# Langchain flow is: template instance builds prompt/messages, model instanec g

chain = chat_template | chat 
```

```python
chain.invoke({'pet':'dog', 'breed':'shepherd'})
```

```python
# Batch method runs invoke() in parallel. Testing Parallelization vs Sequential Calls
%time
chain.batch([{'pet':'dog','breed':'shepherd'},
            {'pet':'dragon','breed':'night fury'}])
```

```python
%time 
chain.invoke({'pet':'dog','breed':'shepherd'})
```

```python
%time 
chain.invoke({'pet':'dragon','breed':'night fury'})
```

```python
response = chain.stream({'pet':'dragon','breed':'night fury'})
```

```python
# Chunks the API Message to Send Responses as Generated
next(response)
```

```python
for i in response:
    print(i.content)
```
