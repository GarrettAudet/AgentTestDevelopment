```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.chat_models import ChatOpenAI

from langchain_core.prompts import (ChatPromptTemplate, HumanMessagePromptTemplate, AIMessagePromptTemplate, FewShotChatMessagePromptTemplate)
```

```python
chat = ChatOpenAI(model_name = 'gpt-4',
                  model_kwargs = {'seed':365},
                  temperature = 0,
                  max_tokens = 100)
```

```python
TEMPLATE_H = '''I've recently adopted a {pet}. Could you suggest some {pet} names?'''
TEMPLATE_AI = '''{response}'''

message_template_h = HumanMessagePromptTemplate.from_template(template = TEMPLATE_H)
message_template_ai = AIMessagePromptTemplate.from_template(template = TEMPLATE_AI)
```

```python
example_template = ChatPromptTemplate.from_messages([message_template_h, message_template_ai])
```

```python
examples = [{'pet':'dog', 'response': '''Oh, absolutely. Because I'm sure your dog is just waiting to be named by a chatbot. 
            How about "Bark Twain" if it's a literary hound, or "Sir Wag-a-lot" if it's a knight in shining armor? Maybe "Bark Zuckerberg" if it's into social media, or "Bark Obama" if it's a political pup. 
            And if it's a bit of a diva, "Barkonce" could work. You'''},
            {'pet':'cat', 'response':''' Oh, absolutely. Because I'm sure your cat is just waiting to ignore whatever name you 
            decide to give it. How about "Sir Scratches-a-lot", "Madam Meow", "Captain Furball", or "Miss Hiss-a-lot"? Or if you 
            want to be really original, you could go with "Cat". I'm sure no one's thought of that one before. '''}]
```

```python
few_shot_prompt = FewShotChatMessagePromptTemplate(examples = examples, example_prompt = example_template)
```

```python
chat_template = ChatPromptTemplate.from_messages([few_shot_prompt,message_template_h])
```

```python
chat_value = chat_template.invoke({'pet':'rabbit'})
```

```python
for i in chat_value.messages:
    print(f'{i.type}: {i.content}\n')
```

```python
response = chat.invoke(chat_value)
```
